```python
import pandas as pd
```

```python
train_df = pd.read_csv('../dados/train_processado.csv')
test_df = pd.read_csv('../dados/test_processado.csv')
```

```python
train_df['Nome'].str.extract(' ([A-Za-z]+)\.', expand=False).value_counts()
```

```python
def extrair_e_tratar_titulo(df):
    # extrai o título da coluna Nome
    df['Titulo'] = df['Nome'].str.extract(' ([A-Za-z]+)\.', expand=False)
    
    # simplifica e agrupa
    df['Titulo'] = df['Titulo'].replace(['Lady', 'Countess', 'Capt', 'Col', 'Don', 
                                        'Dr', 'Major', 'Rev', 'Sir', 'Jonkheer', 'Dona'], 'Outros')
    df['Titulo'] = df['Titulo'].replace(['Mlle', 'Ms'], 'Miss')
    df['Titulo'] = df['Titulo'].replace('Mme', 'Mrs')
    
    return df
```

```python
# aplicando 
train_df = extrair_e_tratar_titulo(train_df)
test_df = extrair_e_tratar_titulo(test_df)
```

```python
train_df = train_df.drop(columns=['Nome'])
test_df = test_df.drop(columns=['Nome'])
```

```python
train_df = pd.get_dummies(train_df, columns=['Titulo'], drop_first=True)
test_df = pd.get_dummies(test_df, columns=['Titulo'], drop_first=True)
```

```python
colunas_booleanas = test_df.select_dtypes(include=['bool']).columns
test_df[colunas_booleanas] = test_df[colunas_booleanas].astype(int)

colunas_booleanas = train_df.select_dtypes(include=['bool']).columns
train_df[colunas_booleanas] = train_df[colunas_booleanas].astype(int)
```

```python
train_df.to_csv('../dados/train_processado.csv', index=False)
test_df.to_csv('../dados/test_processado.csv', index=False)
```
